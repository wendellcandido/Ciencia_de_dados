```python
# Aula de Ciência de Dados
# Prof(a): Eng. Luciana A.

# Exemplo de dados (lista com valores de produção)
producao = [120, 150, 90, 200, 175, 130]

total = sum(producao)
maior = max(producao)
menor = min(producao)
media = total / len(producao)

print("Total:", total)
print("Maior produção:", maior)
print("Menor produção:", menor)
print("Média:", media)
```

```python
# ATIVIDADE

# Definindo os dados da tabela da fábrica
maquinas = ["A", "B", "C", "D", "E"]
producao = [320, 450, 280, 500, 350]

# Questão 1: Quantidade total de peças produzidas
total_pecas = sum(producao)
print(f"1. Total de peças produzidas: {total_pecas}")

# Questão 2: Qual é a maior produção?
maior_producao = max(producao)
print(f"2. Maior produção: {maior_producao}")

# Questão 3: Qual é a menor produção?
menor_producao = min(producao)
print(f"3. Menor produção: {menor_producao}")

# Questão 4: Qual é a média de produção?
media_producao = total_pecas / len(producao)
print(f"4. Média de produção: {media_producao:.2f}")
```

```python
import matplotlib.pyplot as plt

# Questão 5: Gere um gráfico de barras
plt.figure(figsize=(8, 5)) # Define o tamanho da imagem do gráfico
plt.bar(maquinas, producao, color='skyblue', edgecolor='black')

plt.title("Produção Industrial por Máquina", fontsize=14)
plt.xlabel("Máquinas", fontsize=12)
plt.ylabel("Peças Produzidas", fontsize=12)

# Exibe os valores exatos em cima de cada barra para facilitar a leitura
for i in range(len(maquinas)):
    plt.text(i, producao[i] + 10, str(producao[i]), ha='center', fontsize=10)

plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
